```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

figure=plt.figure(figsize=(18,18))
ax=Axes3D(figure)
X=np.arange(-10,10,0.005)
T=np.arange(-10,10,0.005)
X,T=np.meshgrid(X,T)
U=4/((np.exp(np.sqrt(2)/2*(X-2*T))+np.exp(-np.sqrt(2)/2*(X-2*T)))*(np.exp(np.sqrt(2)/2*(X-2*T))+np.exp(-np.sqrt(2)/2*(X-2*T))))
# U=X*T
ax.plot_surface(X,T,U,cmap='summer')
ax.view_init(elev=45,azim=20)
plt.savefig('picture.png',dpi=300)
plt.show()
```
