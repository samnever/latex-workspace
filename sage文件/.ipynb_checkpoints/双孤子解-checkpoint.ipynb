```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

figure=plt.figure(figsize=(18,18))
ax=Axes3D(figure)
x=np.arange(-10,10,0.005)
t=np.arange(-10,10,0.005)
x,t=np.meshgrid(x,t)
u=2*((np.exp(x-t)+4*np.exp(2*x-8*t)+np.exp(3*x-9*t))/(np.exp(x-t)+np.exp(2*x-8*t)+np.exp(3*x-9*t)/9+1)-((np.exp(x-t)+2*np.exp(2*x-8*t)+np.exp(3*x-9*t)/3)/(np.exp(x-t)+np.exp(2*x-8*t)+np.exp(3*x-9*t)/9+1))**2)
# U=X*T
ax.plot_surface(x,t,u,cmap='summer')
# ,cmap='rainbow'
ax.view_init(elev=45,azim=30)
plt.savefig('picture.png',dpi=300)
plt.show()
```
