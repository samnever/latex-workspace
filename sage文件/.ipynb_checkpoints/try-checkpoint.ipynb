```python
var('x k')
x1=pi/2
xmin=x1-pi;xmax=x1+pi
f(x)=sin(x)
p1=plot(f,(xmin,xmax),color='red',legend_label="sin(x)")
t3=f.taylor(x,x1,3)
t5=f.taylor(x,x1,5)
t7=f.taylor(x,x1,7)
p1+=plot(t3,(xmin,xmax),color='blue',legend_label="3")
p1+=plot(t5,(xmin,xmax),color='black',legend_label='5')
p1+=plot(t7,(xmin,xmax),color='purple',legend_label='7')
show(p1)
```
